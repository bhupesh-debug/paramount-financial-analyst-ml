# 01 — Exploratory Data Analysis

All data is synthetic. Monthly spend: 432 rows (36 months × 6 cost centers × opex/capex). CERs: 60 Capital Expenditure Requests. Close entries: 1,675 journal entries across 36 monthly closes.

```python
import pandas as pd
m = pd.read_csv('data/processed/monthly_spend.csv')
c = pd.read_csv('data/processed/cers.csv')
j = pd.read_csv('data/processed/close_entries.csv')
print(m.shape, c.shape, j.shape)
```

```python
m.groupby('cost_center')[['budget','actual']].sum()
```

```python
m.groupby('month')[['budget','actual']].sum().plot(title='Total monthly spend: budget vs actual')
```

```python
c['npv'].describe(), (c['npv'] > 0).mean()
```

```python
j.groupby('close_month')['is_anomaly'].mean().plot(title='Anomaly rate by close month')
```
